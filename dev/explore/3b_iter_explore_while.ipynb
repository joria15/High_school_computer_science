# Conditional Looping

In this last week, we have learned lots about conditionals. We know we can construct an `if` block that checks to see if a `boolean expression` `evaluates` to `True`, and if it does, execute the `body` of the block.

Let's look at an example again:

````python
name = input("Please enter your name: ")

if name.lower() == "habib":
    print("You must be the teacher")
````

This time, there is no `else` (and that's okay). Notice how if the inputted name is not "habib" (ignoring the case), then nothing happens.

## Intro to while loops

Now, we are going to transform this into a `conditional loop` by introducing a new keyword: `while`

**NOTE:** If you input habib, you will need to hit the square stop button to continue this lesson 😅

````python
name = input("Please enter your name: ")

while name.lower() == "habib":
    print("You must be the teacher")
````

One of two things just happened. Either:
- You inputted something other than habib, in which case nothing happened
- You inputted habib, in which case you got caught in an infinite loop

Before I tell you how to fix this, let's look at the anatomy of a while loop:

```python
while some_condition:
    body_of_while_loop
    change_condition
```

Notice that last line. It's important that somewhere in the body of the while loop we change the condition so we have the chance of getting out of the while loop.

Let's modify our code from before just a little bit.

````python
name = input("Please DO NOT enter habib: ")

while name.lower() == "habib":
    print("Incorrect. You must NOT enter habib")
    name = input("Please DO NOT enter habib: ")

print("This code will only run once you enter something other than habib")
````

As you can see, while loops will come in handy when we want to `validate` user input. But let's put that on old for now and do something easier.

What if we want to do something exactly 10 times?

````python
# Begin a counter at 0
counter = 0
# while that counter is less than 10
while counter < 10:
    # Print this line
    print("This code will repeat 10 times")
    # Add one to the counter, and re-save it.
    # So if the counter was 5, we do 5 + 1 = 6, then replace the 5 in counter with 6
    # This is called "incrementing"
    counter = counter + 1
````

## While loops for input validation

Say we want the user to enter a secret password before they are allowed to use your awesome calculator program. Let's set that up together:

````python
# Have the user input two numbers that we will add together
# Input
num1 = int(input("Enter a number: "))
num2 = int(input("Enter a number: "))

# Processing
the_sum = num1 + num2

# Output
print(f"The sum of {num1} and {num2} is {the_sum}")
````

Now, let's look at how we can ask the user to enter a password.

Let's say that the secret password is going to be `hot-chocolate-pizza`

````python
# Have the user input the password, if it's wrong, we won't let them continue

# Input
password = input("Enter the password: ")

# Processing / Output
while password != "hot-chocolate-pizza":
    print("Incorrect password; You shall not pass")
    password = input("Enter the password: ")

print("Calculator unlocked.")
````

Finally, let's put the two pieces together:

````python
# Have the user input the password, if it's wrong, we won't let them continue

# Input
password = input("Enter the password: ")

# Processing / Output
while password != "hot-chocolate-pizza":
    print("Incorrect password; You shall not pass")
    password = input("Enter the password: ")

print("Calculator unlocked.")

# Have the user input two numbers that we will add together

# Input
num1 = int(input("Enter a number: "))
num2 = int(input("Enter a number: "))

# Processing
the_sum = num1 + num2

# Output
print(f"The sum of {num1} and {num2} is {the_sum}")
````
